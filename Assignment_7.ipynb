## Assignment 20: Count Positive, Negative and Zero Values

```python
def count_number_types(number):
    positive_numbers=0
    negative_numbers=0
    zeros=0
    for i in number:
        if i>0:
            positive_numbers+=1
        elif i<0:
            negative_numbers+=1
        else:
            zeros+=1
    return positive_numbers,negative_numbers,zeros
    

number=[10,20,40,5,1,0,0,-52,-8,-1]
positive,negative,zero=count_number_types(number)
print(f"Positive: {positive}")
print(f"Negative: {negative}")
print(f"Zeros: {zero}")
```

## Assignment 33: Function Returning a List

Write a function named `get_even_numbers()`.

### Requirements

- Accept a starting number and ending number.
- Return a list containing all even numbers in the range.
- Validate that the starting number is smaller than the ending number.

```python
def get_even_numbers(starting_number,ending_number):
    even_numbers=[]
    if starting_number>=ending_number:
        print("starting_number must be smaller than ending number")

    for i in range(starting_number,ending_number):
        if i%2==0:
           even_numbers.append(i)
    return even_numbers

starting_number=1
ending_number=10

get_even_numbers(starting_number,ending_number)
```
