## Assignment 34: Prime Number Function

Write a function named `is_prime()`.

### Requirements

- Accept one integer.
- Return `True` when the number is prime.
- Return `False` otherwise.
- Use the function to print all prime numbers from 1 to 100.

### Hint

A prime number has exactly two factors.

```python
def is_prime(n):

    if n < 2:
        return False
    
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
            
    return True

print("Prime numbers from 1 to 100:")
primes = []

for num in range(1, 101):
    if is_prime(num):
        primes.append(num)

print(", ".join(map(str, primes)))   
```

## Assignment 35: Fibonacci Function

### Example Input

    [0, 1, 1, 2, 3, 5, 8]

## Assignment 92: Closure-Based Bank Balance

Write a function named `create_bank_account()`.

### Requirements
- Store the balance in the outer function.
- Define inner functions for deposit, withdrawal, and balance enquiry.
- Use `nonlocal` to update the balance.
- Reject withdrawals greater than the available balance.
- Return all three inner functions.


```python
def create_bank_account(balance):
    def deposit(amount):
        nonlocal balance
        balance += amount
        print("Deposited:", amount)

    def withdraw(amount):
        nonlocal balance
        if amount > balance:
            print("Insufficient balance")
        else:
            balance -= amount
            print("Withdrawn:", amount)

    def check_balance():
        print("Balance:", balance)

    return deposit, withdraw, check_balance

deposit, withdraw, check_balance = create_bank_account(10000)

deposit(2000)
withdraw(3000)
check_balance()
```

## Assignment 93: Function Decorator

Create a decorator named `execution_logger()`.

### Requirements
- The decorator should print a message before the original function executes.
- It should print another message after execution.
- Apply it to an addition function.
- Preserve the original function's return value.


```python
def execution_logger(function):
    def wrapper(*args, **kwargs):
        print("Starting function...")
        result = function(*args, **kwargs)
        print("Function completed.")
        return result

    return wrapper

@execution_logger
def addition(a, b):
    return a + b

result = addition(10, 20)
print("Result:", result)
```

## Assignment 100: Complete Functional Programming Project

Develop a complete **Employee Performance Analysis System** using functions, lambda expressions, `map()`, `filter()`, `reduce()`, nested functions, closures, and LEGB scope.

### Functional Requirements
1. Display all employees.
2. Extract employee names using `map()`.
3. Filter high performers using `filter()`.
4. Search employees by department.
5. Calculate revised salaries using performance scores.
6. Calculate current and revised total salary using `reduce()`.
7. Find the highest-paid employee using `reduce()`.
8. Sort employees using `sorted()` and lambda expressions.
9. Create closure-based 5%, 10%, and 15% bonus calculators.
10. Generate a department report using nested functions.
11. Demonstrate LEGB scope.
12. Create a menu-driven program with options 1-10.


```python
from functools import reduce

company_name = "Edukron"

employees = [
    {
        "id": 101,
        "name": "Ravi",
        "department": "IT",
        "salary": 60000,
        "performance_score": 88,
        "experience": 5
    },
    {
        "id": 102,
        "name": "Anitha",
        "department": "HR",
        "salary": 45000,
        "performance_score": 72,
        "experience": 4
    },
    {
        "id": 103,
        "name": "Kiran",
        "department": "IT",
        "salary": 75000,
        "performance_score": 94,
        "experience": 7
    },
    {
        "id": 104,
        "name": "Meena",
        "department": "Finance",
        "salary": 55000,
        "performance_score": 81,
        "experience": 6
    },
    {
        "id": 105,
        "name": "Arjun",
        "department": "IT",
        "salary": 40000,
        "performance_score": 65,
        "experience": 2
    }
]


# Part 1: Display Employees
def display_employees(data):
    if not data:
        print("No employees found.")
        return

    for employee in data:
        print(
            f"ID: {employee['id']}, "
            f"Name: {employee['name']}, "
            f"Department: {employee['department']}, "
            f"Salary: {employee['salary']}, "
            f"Performance: {employee['performance_score']}, "
            f"Experience: {employee['experience']}"
        )


# Part 2: Extract Employee Names
def get_names(data):
    return list(map(lambda employee: employee["name"], data))


# Part 3: Filter High Performers
def high_performers(data):
    return list(
        filter(lambda employee: employee["performance_score"] >= 80, data)
    )


# Part 4: Filter by Department
def search_department(data, department):
    return list(
        filter(
            lambda employee: employee["department"].lower() == department.lower(),
            data
        )
    )


# Part 5: Salary Increase
def revised_salary(employee):
    score = employee["performance_score"]

    if score >= 90:
        increase = 15
    elif score >= 80:
        increase = 10
    elif score >= 70:
        increase = 5
    else:
        increase = 0

    new_employee = employee.copy()
    new_employee["revised_salary"] = round(
        employee["salary"] + employee["salary"] * increase / 100, 2
    )

    return new_employee


def calculate_revised_salaries(data):
    return list(map(revised_salary, data))


# Part 6: Total Salary Expense
def salary_expense(data):
    current_total = reduce(
        lambda total, employee: total + employee["salary"],
        data,
        0
    )

    revised_data = calculate_revised_salaries(data)

    revised_total = reduce(
        lambda total, employee: total + employee["revised_salary"],
        revised_data,
        0
    )

    additional_expense = revised_total - current_total

    return current_total, revised_total, additional_expense


# Part 7: Highest-Paid Employee
def highest_paid(data):
    if not data:
        return None

    return reduce(
        lambda x, y: x if x["salary"] >= y["salary"] else y,
        data
    )


# Part 8: Sort Employees
def sort_employees(data, choice):
    if choice == "name":
        return sorted(data, key=lambda employee: employee["name"])

    elif choice == "salary":
        return sorted(data, key=lambda employee: employee["salary"])

    elif choice == "performance":
        return sorted(data, key=lambda employee: employee["performance_score"])

    elif choice == "experience":
        return sorted(data, key=lambda employee: employee["experience"])

    return data


# Part 9: Closure-Based Bonus Calculator
def create_bonus_calculator(percentage):
    def calculate_bonus(salary):
        return salary * percentage / 100

    return calculate_bonus


# Part 10 and 11: Department Report + LEGB
def generate_department_report(data, department):
    department_name = department

    def employee_count():
        return len(data)

    def total_salary():
        return reduce(
            lambda total, employee: total + employee["salary"],
            data,
            0
        )

    def average_salary():
        if len(data) == 0:
            return 0
        return total_salary() / len(data)

    def average_performance():
        if len(data) == 0:
            return 0

        total = reduce(
            lambda score, employee: score + employee["performance_score"],
            data,
            0
        )
        return total / len(data)

    report_title = department_name + " Department Report"

    return {
        "company": company_name,
        "title": report_title,
        "department": department_name,
        "employee_count": employee_count(),
        "total_salary": total_salary(),
        "average_salary": round(average_salary(), 2),
        "average_performance": round(average_performance(), 2)
    }


# Part 12: Menu-Driven Program
def employee_system():
    while True:
        print("\n===== Employee Performance Analysis System =====")
        print("1. Display all employees")
        print("2. Display employee names")
        print("3. Display high performers")
        print("4. Search by department")
        print("5. Calculate revised salaries")
        print("6. Calculate total salary expense")
        print("7. Find highest-paid employee")
        print("8. Sort employees")
        print("9. Generate department report")
        print("10. Exit")

        choice = input("Enter your choice: ")

        if choice == "1":
            display_employees(employees)

        elif choice == "2":
            print("Employee names:", get_names(employees))

        elif choice == "3":
            display_employees(high_performers(employees))

        elif choice == "4":
            department = input("Enter department: ")
            result = search_department(employees, department)

            if result:
                display_employees(result)
            else:
                print("Department not found.")

        elif choice == "5":
            result = calculate_revised_salaries(employees)

            for employee in result:
                print(
                    employee["name"],
                    "Current:", employee["salary"],
                    "Revised:", employee["revised_salary"]
                )

        elif choice == "6":
            current, revised, additional = salary_expense(employees)

            print("Current total salary:", current)
            print("Revised total salary:", revised)
            print("Additional salary expense:", additional)

        elif choice == "7":
            employee = highest_paid(employees)

            if employee:
                print("Highest-paid employee:", employee)
            else:
                print("No employees available.")

        elif choice == "8":
            print("1. Name")
            print("2. Salary")
            print("3. Performance")
            print("4. Experience")

            sort_choice = input("Enter sorting choice: ")

            options = {
                "1": "name",
                "2": "salary",
                "3": "performance",
                "4": "experience"
            }

            if sort_choice in options:
                result = sort_employees(employees, options[sort_choice])
                display_employees(result)
            else:
                print("Invalid sorting choice.")

        elif choice == "9":
            department = input("Enter department: ")
            result = search_department(employees, department)

            if result:
                report = generate_department_report(result, department)

                print("\n===== Department Report =====")
                for key, value in report.items():
                    print(key, ":", value)
            else:
                print("Department not found.")

        elif choice == "10":
            print("Program ended.")
            break

        else:
            print("Invalid menu choice.")

employee_system()
```
