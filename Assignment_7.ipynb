## Assignment 8: Area and Circumference of a Circle

Create two functions:

- `circle_area(radius)`
- `circle_circumference(radius)`

### Requirements

- Accept the radius from the user.
- Use `math.pi`.
- Return the area and circumference separately.
- Display both values rounded to two decimal places.

### Formulas

    Area = π × radius²
    Circumference = 2 × π × radius

### Hint

Import the `math` module.

```python
from math import pi
def circle_area(radius):
  return f"Area {pi*(radius**2)}"

def circle_circumference(radius):
  return f"circumference {2*pi*radius}"

radius=int(input("Enter the radius: "))
area=circle_area(radius)
circumference=circle_circumference(radius)
print(area)
print(circumference)
```

## Assignment 9: Student Grade Calculator

Write a function named `calculate_grade()`.

### Requirements

- Accept a student's marks.
- Return the grade according to the following rules:
  - 90–100: A
  - 80–89: B
  - 70–79: C
  - 60–69: D
  - Below 60: F
- Validate that marks are between 0 and 100.

### Example Input

    Enter marks: 84

### Expected Output

    Grade: B

```python
def calculate_grade(student_marks):
  if 100>= student_marks >=90:
    return f"Grade: A"
  elif 89>= student_marks >=80:
    return f"Grade: B"
  elif 79>= student_marks >=70:
    return f"Grade: C"
  elif 69>= student_marks >=60:
    return f"Grade: D"
  elif student_marks <=60:
    return f"Grade: F"


while True:
  student_marks=int(input("Enter marks: "))

  if student_marks <0:
    print("Enter the correct marks")
  elif student_marks >100:
    print("Enter the correct marks")
  else:
    break

Grade=calculate_grade(student_marks)
print(Grade)
```

## Assignment 12: Check Leap Year

Write a function named `is_leap_year()`.

### Requirements

- Accept a year as an integer.
- Return `True` when it is a leap year.
- Otherwise, return `False`.
- Print an appropriate message.

### Rules

A leap year is:

- Divisible by 400, or
- Divisible by 4 but not divisible by 100.

```python
def is_leap_year(year):
  if (year % 4 ==0 and year % 100 !=0) or (year % 400 ==0):
    return True
  else:
    return False

year=int(input("Enter the year: "))
result=is_leap_year(year)

print(result)
```

## Assignment 13: Factorial Using a Function

Write a function named `calculate_factorial()`.

### Requirements

- Accept a non-negative integer.
- Calculate its factorial using a loop.
- Do not use `math.factorial()`.
- Return the factorial.
- Handle negative input.

### Example Input

    Enter a number: 5

### Expected Output

    Factorial: 120

```python
def calculate_factorial(number):
  result=1
  for i in range(1,number+1):
    result *=i
  return result

while True:
  number=int(input("Enter a number: "))

  if number<0:
    print("Enter the correct number")
  else:
    break

factorial=calculate_factorial(number)
print(factorial)
```
